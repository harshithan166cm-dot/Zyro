```python
import numpy as np
from pypdf import PdfReader
from pathlib import Path

# --- RAG Configurations ---
CHUNK_SIZE = 150
CHUNK_OVERLAP = 30
TOP_K = 3
ANSWER_RULES = "Answer the question truthfully using only the provided context."

# --- Dummy / Placeholder Models (Replace with your actual implementations) ---
class DummyEmbedder:
    def encode(self, texts, normalize_embeddings=True):
        # Returns random vectors for demonstration purposes
        num_texts = len(texts)
        vectors = np.random.randn(num_texts, 384)
        norms = np.linalg.norm(vectors, axis=1, keepdims=True)
        return vectors / np.maximum(norms, 1e-12)

embedder = DummyEmbedder()

def generate(prompt, system_text=None, demo_reply=None):
    return demo_reply if demo_reply else "[Model Response Placeholder]"

# --- Knowledge Base and Functions ---
knowledge_base = {"name": None, "chunks": [], "vectors": None}  # populated by build_knowledge_base()


def read_pdf(path):
    """Returns a list of page texts from a PDF file."""
    return [page.extract_text() or "" for page in PdfReader(path).pages]


def split_into_chunks(pages):
    """STAGE 1: split each page into overlapping chunks of about CHUNK_SIZE words."""
    chunks = []
    step = max(1, CHUNK_SIZE - CHUNK_OVERLAP)
    for page_number, text in enumerate(pages, start=1):
        words = text.split()
        for start in range(0, len(words), step):
            chunk_words = words[start:start + CHUNK_SIZE]
            if len(chunk_words) > 5:  # skip very short fragments
                chunks.append({"page": page_number, "text": " ".join(chunk_words)})
            if start + CHUNK_SIZE >= len(words):
                break
    return chunks


def build_knowledge_base(pages, name):
    """STAGE 2: convert each chunk into an embedding."""
    chunks = split_into_chunks(pages)
    if not chunks:
        raise ValueError("No text was found. The PDF may contain scanned images only.")
    vectors = embedder.encode([c["text"] for c in chunks], normalize_embeddings=True)
    knowledge_base.update(name=name, chunks=chunks, vectors=vectors)
    return f"Loaded **{name}**: {len(pages)} pages, {len(chunks)} chunks."


def retrieve(question, top_k=TOP_K):
    """STAGE 3: find the chunks whose meaning is closest to the question."""
    question_vector = embedder.encode([question], normalize_embeddings=True)[0]
    similarities = knowledge_base["vectors"] @ question_vector   # 1.0 = identical meaning
    best = np.argsort(similarities)[::-1][:top_k]
    return [{**knowledge_base["chunks"][i], "score": float(similarities[i])} for i in best]


def answer(question):
    """STAGE 4: give the retrieved chunks to Gemma and ask it to answer using only them."""
    found = retrieve(question)
    context = "\n\n".join(f"[Page {c['page']}] {c['text']}" for c in found)
    prompt = f"Context:\n{context}\n\nQuestion: {question}"
    demo = (f"**Demonstration mode:** Retrieval completed; "
            f"the most relevant passage is on **page {found[0]['page']}**:\n\n> {found[0]['text'][:400]}...")
    reply = generate(prompt, system_text=ANSWER_RULES, demo_reply=demo)
    return reply, found


# Quick test with the sample document.
# Find the PDFs uploaded to Google Colab
pdf_files = sorted(Path("/content").glob("*.pdf"))

if not pdf_files:
    raise FileNotFoundError(
        "No PDFs found. Upload your SSIT PDF files first."
    )

# Read all PDF pages and preserve their source information
SAMPLE_PAGES = []
PAGE_SOURCES = []

for pdf_path in pdf_files:
    pages = read_pdf(str(pdf_path))

    for page_number, page_text in enumerate(pages, start=1):
        if page_text.strip():
            SAMPLE_PAGES.append(
                f"Source: {pdf_path.name}\n"
                f"Page: {page_number}\n"
                f"{page_text}"
            )
            PAGE_SOURCES.append({
                "source": pdf_path.name,
                "page": page_number
            })

print("PDF files found:", len(pdf_files))
print("Pages extracted:", len(SAMPLE_PAGES))

for pdf_path in pdf_files:
    print("-", pdf_path.name)

# Build the knowledge base using your SSIT documents
print(build_knowledge_base(SAMPLE_PAGES, "SSIT Documents"))

# Test retrieval with a question relevant to your documents
test_question = "What information is available in these SSIT documents?"
reply, found = answer(test_question)

print("\nQuestion:", test_question)
print("Answer:", reply)
print("\nRetrieved passages:")

for c in found:
    print(
        f"[Page {c['page']}, similarity {c['score']:.2f}] "
        f"{c['text'][:200]}..."
    )
```

```python

import os
import gradio as gr

gr.close_all()

APP_TITLE = "SSIT Campus Information Assistant"
APP_DESCRIPTION = (
    "Ask questions about SSIT documents. "
    "Answers are based on the loaded PDF and show retrieved page references."
)


def on_upload(file):
    if file is None:
        return "Please upload an SSIT PDF."

    try:
        pages = read_pdf(file)

        if not any(page.strip() for page in pages):
            raise ValueError(
                "No readable text found. The PDF may need OCR."
            )

        return build_knowledge_base(
            pages,
            os.path.basename(file)
        )

    except Exception as e:
        raise gr.Error(f"Could not load the PDF: {e}")


def on_use_sample():
    if not SAMPLE_PAGES:
        raise gr.Error(
            "No sample documents are loaded. Upload an SSIT PDF first."
        )

    try:
        return build_knowledge_base(
            SAMPLE_PAGES,
            "SSIT Sample Documents"
        )
    except Exception as e:
        raise gr.Error(f"Could not load sample documents: {e}")


def on_ask(question):
    if not question or not question.strip():
        raise gr.Error("Please enter a question.")

    if knowledge_base["vectors"] is None:
        raise gr.Error("Upload an SSIT PDF before asking questions.")

    try:
        reply, found = answer(question)

        sources = "\n\n".join(
            f"**{knowledge_base['name']} — Page {c['page']}** "
            f"(similarity: {c['score']:.2f})\n\n"
            f"> {c['text']}"
            for c in found
        )

        return reply, sources

    except Exception as e:
        raise gr.Error(f"Could not answer the question: {e}")


with gr.Blocks(title=APP_TITLE) as demo:
    gr.Markdown(f"# {APP_TITLE}\n{APP_DESCRIPTION}")

    with gr.Row():
        pdf_input = gr.File(
            label="Upload an SSIT PDF",
            file_types=[".pdf"],
            type="filepath"
        )

        with gr.Column():
            status_md = gr.Markdown(
                "Current document: **None loaded**"
            )
            sample_btn = gr.Button("Load prepared SSIT documents")

    question_box = gr.Textbox(
        label="Ask a question",
        placeholder="Example: What are the library timings?"
    )

    ask_btn = gr.Button("Ask", variant="primary")

    gr.Examples(
        examples=[
            "What information is available in this document?",
            "What are the library timings?",
            "What are the attendance requirements?",
            "What are the examination rules?"
        ],
        inputs=question_box
    )

    answer_md = gr.Markdown()

    with gr.Accordion("Retrieved passages and page references", open=False):
        sources_md = gr.Markdown()

    pdf_input.upload(
        on_upload,
        inputs=pdf_input,
        outputs=status_md
    )

    sample_btn.click(
        on_use_sample,
        outputs=status_md
    )

    ask_btn.click(
        on_ask,
        inputs=question_box,
        outputs=[answer_md, sources_md]
    )

    question_box.submit(
        on_ask,
        inputs=question_box,
        outputs=[answer_md, sources_md]
    )

demo.launch(share=True)
```
